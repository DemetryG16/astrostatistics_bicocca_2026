2) One possible model for this data is a sharp burst with an exponential tail rising above a nominal background level. This model can be described as:

$y = b$, for $t<t_0$

$y = b + A\exp[-\alpha(t-t_0)]$, for $t\geq t_0$

where $b$ is the background amplitude, $A$ is the burst amplitude, $t_0$ is the burst epoch, and $\alpha$ encodes how rapidly the burst dies off. Code this burst model,

```python
def y(t,variables):
    b,A,alpha,t_0 = variables
    return np.where(t_0 > t,b, b + A*np.exp(-alpha*(t-t_0)))

ttx = np.linspace(min(time),max(time),1000)    
tty = np.linspace(min(flux),max(flux),1000)

plt.subplots(figsize =(13,5))
plt.plot(ttx, y(ttx,10,max(flux)-10,np.mean(error)/10,np.sort(time)[np.argmax(flux)]))
plt.errorbar(time,flux,yerr=error, fmt='o', color = 'green')
plt.title('Implementing the model');
```

```python
def logLH(variables,sigma,flux,y=y):
    b,A,alpha,t_0 = variables
    y_time = y(time,variables)
    return -0.5*np.sum((flux-y_time)**2 /sigma**2)

# I don't know why the prior is proportional to 1/alpha, check it later?????????
def logprior(alpha):
    return 1-np.log(alpha)

def logposterior(variables,sigma,flux,y, alpha):
    return logLH(variables,sigma,flux,y)*logprior(alpha)
```
